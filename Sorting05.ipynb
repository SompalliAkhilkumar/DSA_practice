Selection sort


```python
a = [5,4,3,2,1]

for i in range(0, len(a)-1):
    mi = i
    for j in range(i+1,len(a)):
        if a[j] < a[mi]:
            mi = j
    a[i],a[mi] = a[mi],a[i]

print(a)
```

Bubble sort

```python
a = [5,4,3,2,1,6,9,7,8]

for i in range(len(a),0,-1):
    for j in range(0,i-1):
        if a[j] > a[j+1]:
            a[j],a[j+1] = a[j+1],a[j]
print(a)
```

Insertion sort

```python
a = [1,2,5,4,3,6,8,7,9]

for i in range(len(a)):
    j = i
    while(j>0 and a[j-1]>a[j]):
        a[j-1],a[j] = a[j],a[j-1]
        j -= 1
print(a)
```

```python
def bubble_sort(a):
    for i in range(len(a),0,-1):
        for j in range(0,i-1):
            if a[j] > a[j+1]:
                a[j],a[j+1] = a[j+1],a[j]
    return a
a = [5,4,3,2,1,6,9,7,8]
print(bubble_sort(a))
```

```python
def insertion_sort(a):
    for i in range(len(a)):
        j = i
        while(j>0 and a[j-1]>a[j]):
            a[j-1],a[j] = a[j],a[j-1]
            j -= 1
    return a
a = [1,2,5,4,3,6,8,7,9]
print(insertion_sort(a))
```

Merge Sort

```python
def msort(a,l,h):
    if l >= h:
        return
    m = (l+h)//2
    msort(a,l,m)
    msort(a,m+1,h)
    merge(a,l,m,h)
```

```python
def merge(a,l,m,h):
    b = []
    i  = l
    j = m+1
    while(i<=m and j<=h):
        if a[i] < a[j]:
            b.append(a[i])
            i += 1
        else:
            b.append(a[j])
            j += 1
    while(i<=m):
        b.append(a[i])
        i += 1
    while(j<=h):
        b.append(a[j])
        j += 1
    for i in range(l,h+1):
        a[i] = b[i-l]
a = [1,2,5,4,3,6,8,7,9]
print(msort(a,0,len(a)-1))
```
