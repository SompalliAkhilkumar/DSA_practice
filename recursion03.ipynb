```python
def f(n):
    if n<1:
        return
    print("a")
    f(n-1)
n = int(input())
f(n)
```

```python
def f(i,n):
    if(i>n):
        return
    print(i)
    f(i+1,n)
f(1,5)
```

```python
def f(i,n):
    if(i<1):
        return
    print(i)
    f(i-1,n)
f(5,5)
```

```python
#back tracking 1 to n
def f(i,n):
    if(i<1):
        return
    f(i-1,n)
    print(i)
f(5,5)
```

```python
#back tracking n to 1
def f(i,n):
    if(i>n):
        return
    f(i+1,n)
    print(i)
f(1,5)
```

```python
#parameterised sum of n
def nadd(n,s):
    if(n<1):
        return s
    return nadd(n-1,s+n)
a = nadd(6,0)
print(a)
```

```python
#functional sum of n
def nadd(n):
    if(n == 0):
        return 0
    return n+nadd(n-1)
a = nadd(6)
print(a)
```

```python
#swap an array using two pointers
def swapn(a,i,n):
    if(i >= n):
        return a
    a[i],a[n] = a[n],a[i]
    return swapn(a,i+1,n-1)
a = [1,2,3,4,5]
b=swapn(a,0,len(a)-1)
print(b)
```

```python
#swap an array using single pointer
def swapn(a,n):
    if(n <= len(a)//2):
        return a
    a[len(a)-n-1],a[n] = a[n],a[len(a)-n-1]
    return swapn(a,n-1)
a = [1,2,3,4,5]
b=swapn(a,len(a)-1)
print(b)
```

```python
#palindrome
def swapn(a,i,n):
    if(i >= n):
        return True
    if (a[i] != a[n]):
        return False
    return swapn(a,i+1,n-1)
a = [1,1,1,2,1]
b=swapn(a,0,len(a)-1)
print(b)
```

```python
#fibonacci
def f(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return f(n-1)+f(n-2)
n = 5
f(n)
```

```python
#factorial
def f(n):
    if(n <= 1):
        return 1
    return n*f(n-1)
n = 3
f(n)
```
