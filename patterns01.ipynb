```python
for i in range(1,6):
    for j in range(1,6):
        if i%2 == 0 and j%2 == 0:
            print(1,end=' ')
        else:
            print(0,end=' ')
    print()
```

```python
n = 5
for i in range(1, n+1):
    for j in range(1, i+1):
        print((i + j + 1) % 2, end=" ")
    print()
```

```python
for i in range(1,5):
    for j in range(1,i+1):
        print(j,end='')
    print(' '*(2*(4-i)),end='')
    for j in range(i,0,-1):
        print(j,end='')
    print()
```

```python
n = 1
for i in range(1,6):
    for j in range(i):
        print(n,end=' ')
        n = n + 1
    print()
```
