```python
a = [1,2,3,4,5,6,7]
n = int(input())
for i in range(len(a)):
    for j in range(len(a)):
        if (i == j):
            continue
        if (a[i] + a[j] == n):
            print(i,j)
```

```python
a = [1,2,3,4,5,6,7]
n = int(input())
for i in range(len(a)):
    for j in range(i,len(a)):
        if (i == j):
            continue
        if (a[i] + a[j] == n):
            print(i,j)
```
