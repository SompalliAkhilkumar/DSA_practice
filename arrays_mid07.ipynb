```python
a = [1,2,3,4,5,6,7]
n = int(input())
for i in range(len(a)):
    for j in range(len(a)):
        if (i == j):
            continue
        if (a[i] + a[j] == n):
            print(i,j)
```

```python
a = [1,2,3,4,5,6,7]
n = int(input())
for i in range(len(a)):
    for j in range(i,len(a)):
        if (i == j):
            continue
        if (a[i] + a[j] == n):
            print(i,j)
```

```python
a = [1,2,3,4,5,6,7]
n = int(input())
for i in range(len(a)):
        x = n - a[i]
        if x in a:
            print(i,a.index(x))
```

```python
a = [1,0,1,0,2,0,2,2,1,2,0,0]
i = a.count(0)
j = a.count(1)
k = a.count(2)
for x in range(i):
    a[x] = 0
for x in range(i,i+j):
    a[x] = 1
for x in range(i+j,i+j+k):
    a[x] = 2
print(a)
```
