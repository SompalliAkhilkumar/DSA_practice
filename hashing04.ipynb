```python
from collections import Counter
a = ['apple', 'banana', 'apple', 'orange', 'banana', 'banana']
freq = Counter(a)
print(dict(freq))
```

```python
a = ['apple', 'banana', 'apple', 'orange', 'banana', 'banana']

freq = {}
for item in a:
    freq[item] = freq.get(item, 0) + 1

print(freq)
```

```python
a = [1,2,3,4,5,4,5,3,2,4,3,5,4,3,2,1,2,3,4,4,4,4]

f = {}
for x in a:
  if x not in f:
    f[x] = 1
  else:
    f[x] += 1
print(f)
```

```python
m = 0
for x in f.values():
  if x > m:
    m = x
print(m)
```
