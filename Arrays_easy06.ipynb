```python
a = [5,6,7,4,3,2,1,9,9,2,6,9,7,8,9,8,9,3,7]
a.sort()
print(a)
l = a[len(a)-1]
for i in range(len(a)-2,0,-1):
    if a[i] != l:
        print(a[i])
        break
```

```python
a = [4,5,3,7,9,7,8,2,9,8,9]
m = 0
s = 0
for i in a:
    if i > m:
        m = i
    if i < m and i > s:
        s = i
print(m)
print(s)
```

```python
a = [2, 3, 7, 8, 9]

is_sorted = all(a[i] <= a[i+1] for i in range(len(a) - 1))
print(is_sorted)
```

```python
a = [4,5,6,7,4,5,6,7,7,8,9,1,2,3,4,0]
b = []
for x in a:
    if x not in b:
        b.append(x)
b
```

```python
a = [2,4,2,3,5,6,5,7,9,4,7,3,6,9]
b = a[0]
for i in range(1,len(a)):
    a[i-1] = a[i]
a.append(b)
a
```

```python
a = [2,4,2,3,5,6,5,7,9,4,7,3,6,9]
k = int(input())
t = a[:k]
b = []
for i in range(k,len(a)):
    b.append(a[i])
b.extend(t)
b
```

```python
a = [2,4,2,3,5,6,5,7,9,4,7,3,6,9]
b = a[:k]
k = int(input())
for i in range(k,len(a)):
    a[i-k] = a[i]
a.extend(b)
a
```

```python
a = [2,4,2,3,5,6,5,7,0,0,0,0,0,9,4,7,3,0,6,9,0]
c = 0
for i in a:
    if i != 0:
        a[c] = i
        c += 1
for i in range(c, len(a)):
    a[i] = 0
a
```
