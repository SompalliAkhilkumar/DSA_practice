```python
a = [5,6,7,4,3,2,1,9,9,2]
a.sort()
print(a)
print(a[len(a)-1])
```

```python
a = [4,5,3,7,9,7,8,2]
m = 0
for i in a:
    if i > m:
        m = i
print(m)
```

```python
a = [5,6,7,4,3,2,1,9,9,2,6,9,7,8,9,8,9,3,7]
a.sort()
print(a)
l = a[len(a)-1]
for i in range(len(a)-2,0,-1):
    if a[i] != l:
        print(a[i])
        break
```

```python
a = [4,5,3,7,9,7,8,2,9,8,9]
m = 0
s = 0
for i in a:
    if i > m:
        m = i
    if i < m and i > s:
        s = i
print(m)
print(s)
```

```python
a = [2, 3, 7, 8, 9]

is_sorted = all(a[i] <= a[i+1] for i in range(len(a) - 1))
print(is_sorted)
```

```python
a = [4,5,6,7,4,5,6,7,7,8,9,1,2,3,4,0]
b = []
for x in a:
    if x not in b:
        b.append(x)
b
```
